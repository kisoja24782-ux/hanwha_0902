```python
from langchain_upstage import UpstageDocumentParseLoader

file_path = "SPRI_AI_Brief_2023년12월호_F.pdf"
```

```python
loader = UpstageDocumentParseLoader(
    file_path,
    output_format="text",
    split="page",
    ocr="auot"
)

docs = loader.load()
```

```python
for doc in docs[:3]:
    print(doc)
```

```python
loader = UpstageDocumentParseLoader(
    file_path,
    output_format="html",
    split="element",
    ocr="auot"
)

docs = loader.load()
```

```python
for doc in docs[:3]:
    print(doc)
```
