```python
from langchain_community.document_loaders import DirectoryLoader

loader = DirectoryLoader(".", glob= "README_test.md")
# 폴더 경로와 glob패턴을 따로 준다.
# 여러 개를 한 번에 읽는 게 이 로더의 용도라 보통 이렇게 사용

docs = loader.load()

len(docs)
```

```python
print(docs[0].page_content[:100])
```

```python
loader = DirectoryLoader(
    "../",glob="*.md", show_progress= True
) # 진행 상태 표시줄이 표시됨 , 문서의 양이 많다면 진행상황을 알 수 있음

docs = loader.load()
```

```python
loader = DirectoryLoader(
    "../", glob="*.md", use_multithreading=True
)

docs = loader.load()
```

- loader_cls 변경하기
- 파일을 로드하는 방식과 파싱 방법을 결정하며, 이를 활용하면 특정 파일 형식에 최적화된 처리가 가능

```python
from langchain_community.document_loaders import TextLoader

loader = DirectoryLoader(".",glob="*.md", loader_cls=TextLoader,loader_kwargs={"encoding": "utf-8"})

docs = loader.load()

print(docs[0].page_content[:100])
```

- 파이썬 파일 로드

```python
from langchain_community.document_loaders import PythonLoader

loader = DirectoryLoader(".",glob="*.py", loader_cls=PythonLoader)

docs = loader.load()

docs
```
