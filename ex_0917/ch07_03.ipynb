### LLM 답변 캐싱하기
- 인메모리캐시
- SQLite캐시

```python
import os

from dotenv import load_dotenv
from langchain_teddynote import logging
from langchain_openai import ChatOpenAI
from langchain_core.prompts import PromptTemplate

load_dotenv()
logging.langsmith("test0914")


print("OpenAI 키 로드됨 : ", bool(os.getenv("OPENAI_API_KEY")))
print("LangSmith 키 로드됨 : ", bool(os.getenv("LANGSMITH_API_KEY")))
print("LangSmith 프로젝트 : ", os.getenv("LANGSMITH_PROJECT"))
```

```python
llm = ChatOpenAI(model="gpt-4o-mini")

prompt = PromptTemplate.from_template("{country}에 대해서 200자 내외로 요약해줘")

chain = prompt | llm
```

```python
%%time

response = chain.invoke({"country":"나이지리아"})
print(response.content)
```

- 인메모리 캐시 사용

```python
# %pip install langchain-community
```

```python
from langchain_classic.globals import set_llm_cache
from langchain_community.cache import InMemoryCache
```

```python
set_llm_cache(InMemoryCache()) # 인메모리 캐시를 사용

response = chain.invoke({"country":"나이지리아"})
print(response.content)
```

```python
%%time

response = chain.invoke({"country":"나이지리아"})
print(response.content)  # 캐싱된 상태에서 동일한 요청하면 응답을 즉시 반환( 캐시에 저장되어 있음 )
```

- SQLite 캐시 설정

```python
from langchain_community.cache import SQLiteCache

if not os.path.exists("cache"): # 캐시 디렉터리(폴더)를 생성
    os.makedirs("cache")

# SQLiteCache 를 사용
set_llm_cache(SQLiteCache(database_path="cache/llm_cache.db"))
```

```python
%%time

response = chain.invoke({"country":"나이지리아"})
print(response.content)
```
