### GPT 모델의 토큰 사용량 확인하기
- 단, 이 기능은 OpenAI API 에 한정

```python
import os

from langchain_teddynote import logging
from dotenv import load_dotenv


load_dotenv()
logging.langsmith("test0914")


print("OpenAI 키 로드됨 : ", bool(os.getenv("OPENAI_API_KEY")))
print("LangSmith 키 로드됨 : ", bool(os.getenv("LANGSMITH_API_KEY")))
print("LangSmith 프로젝트 : ", os.getenv("LANGSMITH_PROJECT"))
```

```python
from langchain_classic.callbacks import get_openai_callback
from langchain_openai import ChatOpenAI

llm = ChatOpenAI(model="gpt-4o-mini")
```

```python
with get_openai_callback() as cb: # 변수 cb에 기록
    result = llm.invoke("나이지리아의 수도는 어디야?")
    print(cb)
```

```python
with get_openai_callback() as cb:
    result = llm.invoke("나이지리아의 수도는 어디야 ?")
    result = llm.invoke("나이지리아의 수도는 어디야 ?")
    print(f"총 사용된 토큰수 : \t\t{cb.total_tokens}")
    print(f"프롬프트에 사용된 토큰수 : \t{cb.prompt_tokens}")
    print(f"답변에 사용된 토큰수 : \t\t{cb.completion_tokens}")
    print(f"호출에 청구도니 금액(USD) : \t{cb.total_cost}")

print(result)
```
