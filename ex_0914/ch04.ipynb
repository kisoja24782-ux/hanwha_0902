```python
from dotenv import load_dotenv

load_dotenv()  # .env 파일 로드
```

```python
from langchain_openai import ChatOpenAI
```

```python
llm = ChatOpenAI(
    temperature=0.1,
    model = "gpt-4o-mini",
)

question = "호주의 수도는 어디인가요 ?"

response = llm.invoke(question)
```

### logprob 로그 프로버블리티 활성화
- 모델의 답변 확신도(Confidence level) :0에 가까울수록 정답에 가까운 확신을 가진다.

```python
# obj 생성
llm_with_logprob = ChatOpenAI(
    temperature=0.1,
    max_completion_tokens= 2048,
    model="gpt-4o-mini",
).bind(logprobs=True)
```

```python
question1 = "호주의 수도는 어디인가요 ?"

response1 = llm_with_logprob.invoke(question1)
```

```python
response1.response_metadata["logprobs"]
```

```python
answer = llm.stream("호주의 아름다운 관광지 10곳과 주소를 알려주세요")
```

```python
for token in answer:
    print(token.content, end="", flush=True)
```
