### LangChainExpressionLanguage  | LCEL
- chain = prompt | model | output_parser
- 여러 구성 요소를 하나의 체인으로 엮어 주는 표현

```python
from dotenv import load_dotenv

load_dotenv()
```

```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import PromptTemplate
from langchain_teddynote.messages import stream_response
```

```python
prompt = PromptTemplate.from_template("{topic} 에 대해 쉽게 설명해주세요")

model = ChatOpenAI(model="gpt-4o-mini", temperature=0.1)

chain = prompt | model
```

```python
chain
```

```python
# input 변수에 딕셔너리 형태로 {topic}과 그에 대한 입력값을 할당하고 invoke() 함수로 답변을 출력
input = {"topic" : "인공지능 모델의 학습 원리"} 
# invoke 메서드를 사용하여 input을 전달해서 AI모델이 생성한 메시지를 반환
print(chain.invoke(input).content)
```

```python
answer = chain.stream(input) # 스트리밍 출력

stream_response(answer)
```

```python
# 프롬프트 템플릿 변수는 여러 개 설정 가능

# prompt를 PromptTemplate 객체로 생성
prompt = PromptTemplate.from_template("{topic} 에 대해 쉽게 {how} 설명해주세요")

model = ChatOpenAI()

chain = prompt | model
```

```python
chain
```

```python
input = {"topic": "인공지능 모델의 학습 원리", "how" : "5살짜리도 이해하기 쉽게 설명해주세요"}
chain.invoke(input).content
```
