### batch() 함수로 일괄 처리하기
- invoke : 동기(Synchronous) : [단일 입력 -> 단일 출력]: 응답의 청크를 스트리밍
- stream : 동기(Synchronous) : [단일 입력 -> 데이터 스트림] : 입력에 대해 체인을 호출
- batch : 동기(Synchronous) : [입력 목록 -> 출력 목록(List)] : 입력 목록에 대해 체인을 호출

```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
```

```python
from dotenv import load_dotenv

# API KEY 정보로드
load_dotenv()
```

```python
model = ChatOpenAI()

prompt = PromptTemplate.from_template("{topic} 에 대하여 3문장으로 설명해줘")

chain = prompt | model | StrOutputParser()
```

```python
print(chain)
```

```python
print(chain.batch([{"topic":"ChatGPT"}, {"topic":"Instagram"}]))
```

```python
answer = chain.batch([{"topic":"ChatGPT"}, {"topic":"Instagram"}])

print(answer)
print(answer[0])
```

```python
chain.batch(
    [
        {"topic":"ChatGPT"},
        {"topic":"Instagram"},
        {"topic":"멀티모델"},
        {"topic":"프로그래밍"},
        {"topic":"머신러닝"}
    ],
    config = {"max_concurrency": 3}
)
```

### batch() 함수로 일괄 처리하기
- ainvoke : 비동기(Asynchronous) : [단일 입력 -> 단일 출력] : 비동기적으로 응답의 청크를 스트리밍
- astream : 비동기(Asynchronous) : [단일 입력 -> 데이터 스트림] : 비동기적으로 입력에 대해 체인을 호출
- abatch : 비동기(Asynchronous) : [입력 목록(List) -> 출력 목록(List)] : 비동기적으로 입력 목록에 대해 체인을 호출
- astream_log : 비동기(Asynchronous) : [단일 입력 -> 로그 및 상태 스트림] : 최종 응답뿐만 아니라 발생하는 중간 단계를 스트리밍

```python
async for token in chain.astream({"topic":"YouTube"}):
    print(token, end="", flush=True)
```

```python
my_process = chain.ainvoke({"topic": "NVDA"})
```

```python
print(await my_process)
```
