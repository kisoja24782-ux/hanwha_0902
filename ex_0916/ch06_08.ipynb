### 날짜 형식 출력 파서 ( DatetimeOutputParser )
- LLM의 출력을 datetime 형식으로 파싱하는 데 사용할 수 있음

```python
import os

from langchain_classic.output_parsers import DatetimeOutputParser
from langchain_core.prompts import PromptTemplate
from langchain_openai import ChatOpenAI
from langchain_teddynote import logging
from dotenv import load_dotenv

load_dotenv()
logging.langsmith("test0914")

llm = ChatOpenAI(model="gpt-4o-mini", temperature=0)

print("OpenAI 키 로드됨 : ", bool(os.getenv("OPENAI_API_KEY")))
print("LangSmith 키 로드됨 : ", bool(os.getenv("LANGSMITH_API_KEY")))
print("LangSmith 프로젝트 : ", os.getenv("LANGSMITH_PROJECT"))
```

```python
output_parser = DatetimeOutputParser()
output_parser.format = "%Y-%m-%d"
```

```python
print(output_parser.get_format_instructions())
```

```python
template = """Answer the user question:

#Format Instructions: {format_instructions}

#Question: {question}

#Answer:"""


prompt = PromptTemplate.from_template(
    template,
    partial_variables={"format_instructions": output_parser.get_format_instructions()}
)

prompt
```

```python
chain = prompt | ChatOpenAI() | output_parser

output = chain.invoke({"question": "Google이 창업한 연도"})
```

```python
output.strftime("%Y-%m-%d") # 결과를 문자열로 변환
```
