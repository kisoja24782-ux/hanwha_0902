### 열거형 출력 파서 ( EnumOutputParser )
-Enum은 열거형이라고도 부르며, 동등한 레벨에 있는 데이터들을 하나의 구조로 묶어 표현하는 방식

-열거형 출력 파서는 언어 모델의 출력을 미리 정의된 열거형 값 중 하나로 파싱하는 도구

```python
import os 

from enum import Enum
from langchain_classic.output_parsers.enum import EnumOutputParser # classic 추가
from langchain_core.prompts import PromptTemplate # core 추가
from langchain_openai import ChatOpenAI
from langchain_teddynote import logging
from dotenv import load_dotenv

load_dotenv()
logging.langsmith("test0914")

llm = ChatOpenAI(model="gpt-4o-mini", temperature=0)

print("OpenAI 키 로드됨 : ", bool(os.getenv("OPENAI_API_KEY")))
print("LangSmith 키 로드됨 : ", bool(os.getenv("LANGSMITH_API_KEY")))
print("LangSmith 프로젝트 : ", os.getenv("LANGSMITH_PROJECT"))
```

```python
class Colors(Enum):
    RED = "빨간색"
    GREEN = "초록색"
    BLUE = "파란색"
```

```python
Colors.RED
```

```python
parser = EnumOutputParser(enum=Colors)

parser.get_format_instructions()
```

```python
prompt = PromptTemplate.from_template(
    """What color is the following object ?
    
    Object : {object}
    
    Instructions: {instructions}"""
).partial(instructions= parser.get_format_instructions())


chain = prompt | ChatOpenAI() | parser
```

```python
response = chain.invoke({"object":"사과"})
print(response)
```
