### 쉼표로 구분ㄴ된 리스트 출력 파서( CommaSeparatedListOutputParser )
- 쉼표로 구분된 항목을 리스트 형식으로 처리하는 파서

```python
import os

from langchain_core.output_parsers import CommaSeparatedListOutputParser
from langchain_core.prompts import PromptTemplate
from langchain_openai import ChatOpenAI
from dotenv import load_dotenv
from langchain_teddynote import logging
```

```python
llm = ChatOpenAI(model="gpt-4o-mini", temperature=0)

load_dotenv()

print("OpenAI 키 : ", os.getenv("OPENAI_API_KEY")[:8] + "...")
print("LANGSMITH 키 : ", os.getenv("LANGSMITH_API_KEY")[:8] + "...")
print("LangSmith 프로젝트 : ", os.getenv("LANGSMITH_PROJECT"))

logging.langsmith("test0914")
```

```python
output_parser = CommaSeparatedListOutputParser()

format_instructions = output_parser.get_format_instructions()
```

```python
print(format_instructions)
```

```python
prompt = PromptTemplate(
    template= "List five {subject}.\n{format_instructions}",
    input_variables=["subject"],
    partial_variables={"format_instructions": format_instructions}
)

```

```python
model = ChatOpenAI(temperature=0)

chain = prompt | model | output_parser
```

```python
answer = chain.invoke({"subject": "대한민국 관광명소"})

answer
```

```python
for s in chain.stream({"subject": "대한민국 관광명소"}):
    print(s)


# 출력 파서를 스트리밍 출력하면 토큰 단위로 별개의 리스트가 되어 구분되므로 활용성이 떨어짐
```
