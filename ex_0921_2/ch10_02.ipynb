### 문자 단위로 재귀적으로 분할하기

```python
with open("appendix-keywords.txt",encoding= "utf-8") as f:
    file = f.read()
```

```python
print(file[:500])
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

text_splitter = RecursiveCharacterTextSplitter(
    chunk_size = 250,
    chunk_overlap = 50,     # 청크 간의 중복되는 문자 수
    length_function = len,    # 문자열 길이를 제한하는 함수
    is_separator_regex=False  # 구분자로 정규식 사용 여부
)
```

```python
texts = text_splitter.create_documents([file])

print(texts[0]) # 분할된 문서의 첫 번째 문서를 출력
print("====" * 20)
print(texts[1]) # 분할된 문서의 두 번째 문서를 출력
```

```python
text_splitter.split_text(file)[:2] # 텍스트를 분할하여 문자열 리스트로 반환
```
