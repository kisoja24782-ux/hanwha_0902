## 텍스트 분할

### 1. 문자 단위로 분할하기

```python
with open("appendix-keywords.txt",encoding="utf-8") as f:
    file = f.read()
```

```python
print(file[:500])
```

```python
from langchain_text_splitters import CharacterTextSplitter

text_splitter = CharacterTextSplitter(
    separator="\n\n",
    chunk_size=210,
    chunk_overlap = 0,
    length_function = len
)
```

```python
texts = text_splitter.create_documents([file])
print(len(texts[0].page_content))
print(texts[0])
```

```python
metadatas = [
    {"document" : 1},
    {"document" : 2}
]
documents = text_splitter.create_documents(
    [
        file,
        file
    ],
    metadatas= metadatas
)

print(documents[0])
```
