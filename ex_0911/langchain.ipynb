```python
from pathlib import Path
from dotenv import dotenv_values
from openai import OpenAI
```

```python
env_path = Path.cwd() / "key_info.env"
print("키 파일 경로:", env_path)

if not env_path.is_file():
    raise FileNotFoundError(f"파일이 없습니다: {env_path}")

print("✅ 파일 확인 완료")
```

```python
# 셀 재실행 시 이전 클라이언트가 남지 않도록 초기화
client = None

api_key = dotenv_values(
    env_path, encoding="utf-8-sig"
).get("OPENAI_API_KEY")

if not api_key or not api_key.strip():
    raise ValueError("OPENAI_API_KEY 값이 없거나 비어 있습니다.")

client = OpenAI(api_key=api_key.strip(), timeout=30.0)
print("✅ 키 읽기 및 클라이언트 준비 완료")
```

```python
if client is None:
    raise RuntimeError("셀 3을 먼저 정상적으로 실행하세요.")

try:
    response = client.responses.create(
        model="gpt-5-mini",
        input="'연결 성공!'이라고 짧게 답해주세요.",
    )

    print("응답 상태:", response.status)
    print("응답:", response.output_text)

except Exception as e:
    print("❌ 테스트 실패:", e)
```
