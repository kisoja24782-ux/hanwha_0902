```python
from dotenv import dotenv_values
from langchain_openai import ChatOpenAI

keys = dotenv_values("Gemini_API_Key.env", encoding="utf-8-sig")
api_key = keys.get("GEMINI_API_KEY")

if not api_key or not api_key.strip():
    raise ValueError("GEMINI_API_KEY를 찾을 수 없습니다.")



# try:
#     response = llm.invoke("대한민국의 수도는 어디인가요?")
#     print("✅ Gemini 연결 성공")
#     print(response.content)
# except Exception as e:
#     print("❌ 테스트 실패:", e)

```

```python

llm = ChatOpenAI(
    api_key=api_key.strip(),  # Gemini 키
    base_url="https://generativelanguage.googleapis.com/v1beta/openai/",
    model="gemini-3.8-flash",  # Gemini 모델
    temperature= 2.0
)
```

```python

try:
    response = llm.invoke("대한민국의 광역시는 몇개 인가요?")
    print("✅ Gemini 연결 성공")
    print(response.content)
except Exception as e:
    print("❌ 테스트 실패:", e)
```
